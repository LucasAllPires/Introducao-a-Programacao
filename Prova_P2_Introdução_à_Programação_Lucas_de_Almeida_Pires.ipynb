```python
# 02) O Centro Universitário IESB utiliza o sistema de menções para a avaliação do rendimento acadêmico dos estudantes. Considere os critérios a seguir que servem para automatizar a atribuição da menção correspondente
# com base numa nota final:
# • Nota >= 9 → menção “SS”
# • 7 <= Nota < 9 → menção “MS”
# • 5 <= Nota < 7 → menção “MM”
# • 3 <= Nota < 5 → menção “MI”
# • 0 <= Nota < 3 → menção “II”
# Desenvolva um programa que dada uma nota ele indicará a menção correta.

print("IESB - CLASSIFICAÇÃO DA NOTA FINAL")

nome_aluno = str(input("Digite o nome do aluno : "))
nota_aluno = float(input("Digite a nota final do aluno: "))

if nota_aluno > 10 or nota_aluno < 0:
  print("Nota inválida! Digite uma nota de 0 a 10.")

elif nota_aluno >= 9:
  print(f"O aluno {nome_aluno} obteve a menção SS (nota: {nota_aluno})")

elif nota_aluno >= 7:
  print(f"O aluno {nome_aluno} obteve a menção MS (nota: {nota_aluno})")

elif nota_aluno >= 5:
  print(f"O aluno {nome_aluno} obteve a menção MM (nota: {nota_aluno})")

elif nota_aluno >= 3:
  print(f"O aluno {nome_aluno} obteve a menção MI (nota: {nota_aluno})")

else:
  print(f"O aluno {nome_aluno} obteve a menção II (nota: {nota_aluno})")

```

```python
# 04) Considere a seguinte lista de notas em um trecho de código em Python. Calcule quantos alunos obtiveram nota suficiente para aprovação (nota maior ou igual a 5.0)

notas = [4.5, 7.0, 3.0, 8.5, 5.0, 9.0, 7.0, 9.5, 10.0, 9.5]

print(f"Quantidade total de alunos: {len(notas)}")

aprovados = 0

for nota in notas:
  if nota >= 5:
    aprovados += 1

print(f"Quantidade de alunos aprovados: {aprovados}")
print(f"Quantidade de alunos reprovados: {len(notas) - aprovados}")
```

```python
# 05) Implemente um programa em Python que leia um valor inicial N e imprima a sequência de valores do cálculo de N! (N fatorial) e o seu resultado. Use N=5 para testar o seu algoritmo.
# Exemplo: 5! = 5 x 4 x 3 x 2 x 1 = 120

# Usei IA para melhorar a formatação ("end")

print("CÁLCULO DE FATORIAL")

numero = int(input("Digite um número e vamos calcular o seu fatorial: "))

fatorial = 1

print(f"{numero}! = ", end="")

for i in range(numero, 0, -1):
    fatorial *= i

    if i == 1:
        print(f"{i}", end="")
    else:
        print(f"{i} x ", end="")

print(f" = {fatorial}")
```

```python
 # 06) Faça um programa Python que imprime a quantidade de números pares de 100 até 200, incluindo-os.

lista_pares = []

for i in range(100, 201):
  if i % 2 == 0:
    lista_pares.append(i)

print(f"A quantidade de números pares de 100 até 200 é igual a: {len(lista_pares)}")
print(f"Os números pares de 100 até 200 são: \n{lista_pares}")
```

```python
# 07) Faça um programa Python que calcule a soma dos 100 primeiros números inteiros maiores que 1 cujas raízes quadradas são exatas.

print("SOMA DOS 100 PRIMEIROS QUADRADOS PERFEITOS (MAIORES QUE 1)")

soma = 0
quantidade = 0
numero_base = 2

while quantidade < 100:

    quadrado_perfeito = numero_base ** 2
    soma += quadrado_perfeito

    quantidade += 1
    numero_base += 1

print(f"A soma dos 100 primeiros quadrados perfeitos maiores que 1 é: {soma}")
```
