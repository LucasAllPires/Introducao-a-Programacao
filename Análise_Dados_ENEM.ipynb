```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('/content/Enem_2024_Amostra_Perfeita.xlsx')
display(df)
```

```python
display(df.columns)
```

```python
df_sexo = df[df['TP_Sexo'] == "Feminino"]
display(df_sexo)
```

```python
df_distritofederal = df[df['Nome_UF_Prova'] == "Distrito Federal"]
display(df_distritofederal)
```

```python
contagem_cor = df["TP_COR_RACA"].value_counts()
display(contagem_cor)

plt.figure(figsize = (10, 6))
contagem_cor.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Quantidade de inscritos por raça/cor", fontsize=14)
plt.xlabel("Raça/Cor", fontsize=12)
plt.ylabel("Quantidade", fontsize=12)
plt.xticks(rotation=45)


plt.show()
```

```python
notas_matematica = df[df['NOTA_MT_MATEMATICA'] > 680]
tabela_frequencia_notas = notas_matematica['SG_UF_PROVA'].value_counts()

display(tabela_frequencia_notas)

plt.figure(figsize=(10, 6))
tabela_frequencia_notas

tabela_frequencia_notas.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title("Frequência de Notas de Matemática maiores que 680 por UF", fontsize=14, fontweight='bold')
plt.xlabel("UF", fontsize=12)
plt.ylabel("Frequência", fontsize=12)
plt.xticks(rotation=45)

plt.show()
```
